```python
import pandas as pd
from pathlib import Path
```

## Carga y forma general
Dimensiones, tipos y nulos.

```python
RAW_DIR = Path.cwd().parent / "data" / "raw"

df_ref = pd.read_csv(RAW_DIR / "underlyings_reference.csv")
print(df_ref.head())
print("-" * 50)
print(df_ref.info())
print("-" * 50)
print(df_ref.shape)
print("-" * 50)
```

## Unicidad de underlying
Confirma que cada ticker aparece una sola vez.

```python
print(df_ref["underlying"].duplicated().sum())
```

## Volatilidad estructural
Rango de structural_base_vol.

```python
print(df_ref["structural_base_vol"].describe())
```

## Cobertura frente a rfqs.csv
Comprueba que todos los tickers usados en las cestas de rfqs.csv existen en la tabla de referencia.

```python
df_rfqs_raw = pd.read_csv(RAW_DIR / "rfqs.csv")

underlyings_en_rfqs = set(
    ticker
    for basket_str in df_rfqs_raw["underlyings"]
    for ticker in basket_str.split("|")
)
underlyings_en_referencia = set(df_ref["underlying"].unique())

print(underlyings_en_rfqs - underlyings_en_referencia)
```

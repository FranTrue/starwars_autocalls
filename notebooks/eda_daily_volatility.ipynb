```python
import pandas as pd
from pathlib import Path
```

## Carga, cobertura y duplicados
Forma general, tickers distintos, rango de fechas por subyacente y duplicados de (date, underlying).

```python
RAW_DIR = Path.cwd().parent / "data" / "raw"

df_vol = pd.read_csv(RAW_DIR / "daily_volatility.csv")
print(df_vol.head())
print(df_vol.info())
print(df_vol.shape)
print("-" * 50)
print(df_vol["underlying"].nunique())
print(df_vol.groupby("underlying")["date"].agg(["min", "max", "count"]))
print("-" * 50)
print(df_vol.duplicated(subset=["date", "underlying"]).sum())
print("-" * 50)
print(sorted(df_vol["underlying"].unique()))
```
